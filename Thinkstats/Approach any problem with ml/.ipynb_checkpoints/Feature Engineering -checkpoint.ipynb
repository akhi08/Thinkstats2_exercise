```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Generate 2000 random dates between 2016-01-01 and 2019-12-31
dates = pd.date_range(start="2016-01-01", end="2019-12-31", freq="D")
random_dates = np.random.choice(dates, 2000)

# Create 200 unique customers, repeat them
customer_ids = np.random.choice(range(10000, 10200), size=2000, replace=True)

# Categorical features (small integer ranges)
cat1 = np.random.randint(1, 6, size=2000)   # values 1–5
cat2 = np.random.randint(1, 4, size=2000)   # values 1–3
cat3 = np.random.randint(0, 2, size=2000)   # values 0–1

# Numeric feature (normally distributed)
num1 = np.random.normal(loc=0, scale=1, size=2000)

# Assemble DataFrame
df = pd.DataFrame({
    "date": random_dates,
    "customer_id": customer_ids,
    "cat1": cat1,
    "cat2": cat2,
    "cat3": cat3,
    "num1": num1
})

# Sort by date for readability
df = df.sort_values("date").reset_index(drop=True)

print(df.head(10))
print(df.shape)
```

```python
df
# we see that we have a date column, and we can easily extract features
# like the year, month, quarter, etc. from that. Then we have a customer_id column
# which has multiple entries, so a customer is seen many times (not visible in the
# screenshot). And each date and customer id has three categorical and one numerical
# feature attached to it. 
```

There are a bunch of features we can create from it:
- What’s the month a customer is most active in
- What is the count of cat1, cat2, cat3 for a customer
- What is the count of cat1, cat2, cat3 for a customer for a given week of the
year
- What is the mean of num1 for a given customer
- And so on.

```python
def generate_features(df):
    # create bunch of features using the date column
    df.loc[:,'year']=df['date'].dt.year
    df.loc[:,'weekofyear']=df['date'].dt.isocalendar().week
    df.loc[:,'month']=df['date'].dt.month
    df.loc[:,'dayofweek']=df['date'].dt.dayofweek
    df.loc[:,'weekend']=(df['date'].dt.weekday>=5).astype(int)

    #create an aggregate dictionary
    aggs={}
    #for aggregation by month,we calculate the number of unique month values and also the mean
    aggs['month']=['nunique','mean']
    aggs['weekofyear']=['nunique','mean']
    # we aggregate  by num1 and calculate sum,max,min and mean values of this column
    aggs['num1']=['sum','max','min','mean']
    # for customer_id,we calculate the total count
    aggs['customer_id']=['size']
    # again for customer_id,we calculate the total unique
    aggs['customer_id']=['nunique']
    # we group by customer_id and calculate the aggregates
    agg_df=df.groupby('customer_id').agg(aggs)
    agg_df=agg_df.reset_index()
    return agg_df
    
```

```python
generate_features(df)
# Now, we can join this dataframe in figure 2 with the original dataframe with
# customer_id column to start training a model
```

##### Month a customer is most active in

Let’s break it down step by step:
df.groupby('customer_id')['month']
Groups the dataframe by customer_id.
For each customer, we select their month values.
- .agg(lambda x: ...)
Applies a custom aggregation function to the grouped month values.

Here, x is a Pandas Series containing the month values for a single customer_id.
- x.value_counts().index[0]
- x.value_counts() counts how many times each month occurs for that customer.
- .index[0] picks the most frequent month (mode).

- import pandas as pd

- df = pd.DataFrame({'customer_id': [1,1,1,2,2,3,3,3,3],'month': ['Jan','Feb','Jan','Mar','Mar','Apr','Apr','May','Apr']})

result = df.groupby('customer_id')['month'].agg(lambda x: x.value_counts().index[0])

```python
import numpy as np
x = np.array([10, 15, 20, 25, 30, 40, 50, 60, 100])
feature_dict={}
#calculate mean
feature_dict['mean']=np.mean(x)
#calculate max
feature_dict['mean']=np.max(x)
#calucalte min
feature_dict['mean']=np.min(x)
#calculate std
feature_dict['mean']=np.std(x)
#cal. variance
feature_dict['mean']=np.var(x)
# cal. peak to peak
feature_dict['mean']=np.ptp(x)
# percentiles
feature_dict['percentile_10']=np.percentile(x,10)
feature_dict['percentile_60']=np.percentile(x,60)
feature_dict['percentile_90']=np.percentile(x,90)
#quantiles
feature_dict['quantile_5']=np.percentile(x,0.05)
feature_dict['quantile_95']=np.percentile(x,0.95)
feature_dict['quantile_99']=np.percentile(x,0.99)
```

### Time Series Data
- library->tsfresh

```python
# !pip install tsfresh
from tsfresh.feature_extraction import feature_calculators as fc
# tsfresh based features
feature_dict['abs_energy']=fc.abs_energy(x)
feature_dict['count_above_mean']=fc.count_above_mean(x)
feature_dict['count_below_mean']=fc.count_below_mean(x)
feature_dict['mean_abs_change']=fc.mean_abs_change(x)
feature_dict['mean_change']=fc.mean_change(x)
```

```python
print(feature_dict)
```

1. fc.abs_energy(x)

Computes the absolute energy of the time series.

Formula:

\text{abs_energy} = \sum_{i=1}^{n} x_i^2

Example: For x = [1, 3] → 1² + 3² = 10.

Interpretation: A measure of signal strength (larger values = stronger signals).

2. fc.count_above_mean(x)

Counts how many values in x are greater than the mean.

Example: For x = [1, 3, 5], mean = 3.
Values above mean = [5] → count = 1.

Interpretation: Shows skewness towards high values.

3. fc.count_below_mean(x)

Counts how many values in x are less than the mean.

Example: For x = [1, 3, 5], mean = 3.
Values below mean = [1] → count = 1.

Interpretation: Skewness towards low values.
